*   **Pregunta 2.**  

Busca e imprime todas las palabras que terminan con dos o más signos de admiración seguidos, por ejemplo "!!!". 

Debes imprimir tanto la palabra como la totalidad de signos de admiración que le siguen. 

Indica cuántos resultados obtuviste.



```python
#\b\w+ → captura la palabra
#!{2,} → asegura dos o más signos de admiración consecutivos
#re.findall() → devuelve todas las coincidencias

patron = r'\b\w+!{2,}'
resu = []

for doc in docs:
    matches = re.findall(patron, doc)
    resu.extend(matches)

print(resu)
print("Total:", len(resu))

```

*   **Pregunta 3.**  

Busca e imprime todas las palabras que están escritas totalmente en mayúsculas. Cada coincidencia debe ser una sola palabra.

Indica cuántas palabras encontraste.



```python
#\b → límite de palabra
#[A-ZÁÉÍÓÚÑ] → letras mayúsculas
#{2,} → mínimo 2 letras
#\b → termina la palabra
#Uso {2,} para evitar contar letras sueltas como I o A.

patron = r'\b[A-ZÁÉÍÓÚÑ]{2,}\b'
resu = []

for doc in docs:
    matches = re.findall(patron, doc)
    resu.extend(matches)

print(resu)
print("Total:", len(resu))
```

*   **Pregunta 4.**  

Busca e imprime los comentarios en donde todos los caracteres alfabéticos (letras) están en mayúsculas. 

Cada coincidencia encontrada debe ser todo el comentario/enunciado.

Indica cuántos resultados obtuviste.


```python

#[A-Za-zÁÉÍÓÚÑáéíóúñ] → letras


patron = r'[A-Za-zÁÉÍÓÚÑáéíóúñ]'
resu = []

for doc in docs:
    matches = re.findall(patron, doc)

    if matches and all(match.isupper() for match in matches):
        resu.append(doc.strip())

print(resu)
print("Total:", len(resu))
```

*   **Pregunta 6.**  

Busca e imprime todas las cantidades numéricas monetarias, enteras o con decimales, que inician con el símbolo $\$$. 

Indica cuántos resultados obtuviste.

```python
#\$ coinciden con $
#\d{1,9} numero que contenga digitos entre 1 - 9

#(?:,\d{3})*
#(?: ... ) = grupo no capturante (solo agrupa, no guarda)
#,\d{3} = coma seguida de exactamente 3 dígitos
#* = 0 o más repeticiones

#(?:\.\d+)?
#\. = punto decimal
#\d+ = uno o más dígitos
#? = opcional


patron = r'\$\d{1,9}(?:,\d{3})*(?:\.\d+)?'

resu = []

for doc in docs:
    matches = re.findall(patron, doc)
    resu.extend(matches)

print(resu)
print("Total:", len(resu))
```

*   **Pregunta 7.**  

Busca e imprime todas las palabras que sean variantes de la palabra "love", sin importar si incluyen mayúsculas o minúsculas, o la manera en que esté conjugada o alguna otra variación que se haga con dicha palabra.

Indica cuántos resultados obtuviste.

```python
#\b → word boundary
#lov → root of the word
#\w* → any continuation
#Final \b → end of word


patron = r'\blov\w*\b'
resu = []

for doc in docs:
    matches = re.findall(patron, doc, re.IGNORECASE)
    resu.extend(matches)

print(resu)
print("Total:", len(resu))
```

*   **Pregunta 8.**  

Busca e imprime todas las palabras, variantes de "so" y "good", que tengan dos o más "o" en "so" y 3 o más "o" en good. 

Indica cuántas encontraste.


```python
#Parte 1 → \bso{2,}\b
#Parte 2 → \bgo{3,}d\b
#| → OR


patron = r'\bs+o{2,}\b|\bg+o{3,}d+\b'
resu = []

for doc in docs:
    matches = re.findall(patron, doc, re.IGNORECASE)
    resu.extend(matches)

print(resu)
print("Total:", len(resu))
```

*   **Pregunta 9.**  

Busca e imprime todas las palabras que tengan una longitud mayor estrictamente a 10 caracteres alfabéticos.

No se consideran los signos de puntuación o caracteres especiales en la longitud de estas cadenas, solo caracteres alfabéticos en mayúsculas o minúsculas.

Indica la cantidad de palabras encontradas.


```python
#\b word start
#[A-Za-z] charachters allowed in the string
#{11,} qty of charachters
#\b end of word
#this logic includes duplicates

patron = r'\b[A-Za-z]{11,}\b'
resu = []
resu2 =set()

for doc in docs:
    matches = re.findall(patron, doc)
    resu.extend(matches)
    resu2.update(matches)

print(resu)
print("Total con duplicados:", len(resu))
print(resu2)
print("Total sin duplicados:", len(resu2))
```

*   **Pregunta 10.**  

Busca e imprime todas las palabras que inician con una letra mayúscula y terminan con una minúscula, pero que además no sea la primera palabra del comentario/string. 

Indica la cantidad de resultados obtenidos.

```python
#Empiecen con mayúscula → A-Z
#Terminen con minúscula → a-z
#NO sean la primera palabra del comentario

patron = r'\b[A-Z][a-z]+\b'
resu = []

for doc in docs:
    words = doc.split() #separa las palabras de cada comenario
    
    for i, word in enumerate(words):
        if i==0: 
            continue #ignora la primera palabra de cada comentario

        #Limpiar puntuacion
        word_limpia = re.sub(r'[^A-Za-z]', '', word) # limpia espacios y caracteres.

        if re.fullmatch(patron, word_limpia):
            resu.append(word_limpia)

print(resu)
print("Total:", len(resu))
```

*   **Pregunta 11.**  

Busca e imprime la secuencia de dos o más palabras que están separadas por un guion, "-", sin que tengan espacios en blanco entre ellas.

Por ejemplo "Go-Kart" sería válido, pero "Go  -Kart" o "Go  -  Kart" no lo serían. 

Indica la cantidad de resultados obtenidos.

```python
#\b → inicio de palabra
#[A-Za-z]+ → primera palabra
#(?: ... )+ → grupo repetido (al menos 1 vez)
#-[A-Za-z]+ → guion seguido de otra palabra
#\b → fin de palabra
#2 palabras (Go-Kart)
#3+ palabras (high-quality-product)

patron = r'\b[A-Za-z]+(?:-[A-Za-z]+)+\b'
resu = []

for doc in docs:
    matches = re.findall(patron, doc)
    resu.extend(matches)

print(resu)
print("\nTotal:", len(resu))
```

*   **Pregunta 12.**  

Busca e imprime todas las palabras que terminan en "ing" o "ed". 

Indica la cantidad de palabras que encontraste de cada una.

```python
#\b → límite de palabra
#[A-Za-z]+ → palabra
#(?:ing|ed) → termina en "ing" o "ed"
#(?:ing) → termina en "ing"
#(?:ed) → termina en "ed"
#\b → fin de palabra


patron1 = r'\b[A-Za-z]+(?:ing)\b'
patron2 = r'\b[A-Za-z]+(?:ed)\b'
resu1 = []
resu2 = []

for doc in docs:
    matches1 = re.findall(patron1, doc)
    matches2 = re.findall(patron2, doc)

    resu1.extend(matches1)
    resu2.extend(matches2)

print(resu1)
print("\nTotal 'ing':", len(resu1))
print(resu2)
print("\nTotal 'ed':", len(resu2))
```

#**Parte 3. Proceso de limpieza.**

*   **Pregunta 13.**  

Ahora realiza un proceso de limpieza del corpus que incluya los siguientes procesos:

*   Solo se deben considerar caracteres alfabéticos. Es decir, se eliminan todos los signos de puntuación y caracteres especiales.
*   Todos los caracteres alfabéticos se transforman a minúsculas.
*   Se deben eliminar todos los espacios en blanco adicionales que se puedan encontrar en cada comentario.

Al finalizar dicho proceso de limpieza, imprime el resultado de los primeros 10 comentarios resultantes. 
   




```python
docs_limpio = []

for doc in docs:
    #1. Dejar solo letras y espacions
    #[^A-Za-z\s]
    #^ → negación
    #elimina todo excepto:
    #letras
    #espacios
    
    limpio = re.sub(r'[^A-Za-z\s]', '',doc)

    #2. Convertir a minusculas
    #lower - convierte todo a minusculas
    limpio = limpio.lower()

    #3. Eliminamos espacios en blanco
    #.strip(), elimina espacios al inicio y asl final
    limpio = re.sub(r'\s+', ' ', limpio).strip()

    docs_limpio.append(limpio)


print(docs[0:10])

print(docs_limpio[0:10])
```

*   **Pregunta 14.**  

Con el resultado de la limpieza obtenido en la pregunta anterior, realiza ahora un proceso de tokenización por palabras del corpus. 

Es decir, al final de este proceso de tokenización, debes tener como resultado una lista de listas, donde cada comentario estará tokenizado por palabras.

Al terminar calcula el total de tokens obtenido en todo el corpus. 

```python
tokens_corpus = []

for doc in docs_limpio:
    tokens = doc.split() #tokenizacion por palabra
    tokens_corpus.append(tokens)

#imprimimos primeros 10 para validar
print(tokens_corpus[0:10])

#total de tokens en todo el corpus
total_tokens = sum(len(tokens) for tokens in tokens_corpus)
    
print("Total Tokens:", total_tokens)
    


```

*   **Pregunta 15.**  

Finalmente, en este ejercicio definiremos nuestro conjunto de palabras "stopwords", las cuales deberás eliminar de todo el corpus.

```python
tokens_corpus_filtrado =[]

for tokens in tokens_corpus:
    tokens_filtrado = [t for t in tokens if t not in mis_stopwords]
    tokens_corpus_filtrado.append(tokens_filtrado)

#total de tokens
total_tokens_filtrados = sum(len(tokens) for tokens in tokens_corpus_filtrado)

#vocabulario unico
vocabulario = list(set(t for tokens in tokens_corpus_filtrado for t in tokens))
#set elimina duplicados
#list los mantenemos para poder utilizar funciones de listas
vocabulario.sort()

#imprimiendo 

print("Total tokens : ", total_tokens )
print("Total tokens sin stopwords: ", total_tokens_filtrados )
print("Total tokens unicos (vocabulario): ", len(vocabulario))
```
